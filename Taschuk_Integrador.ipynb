````python
# Creá una imagen sintética simple (ej: gradiente o patrón)
# Probá tus funciones con esta imagen
# Crear una imagen sintética tipo gradiente horizontal
def crear_imagen_gradiente(ancho=256, alto=256):
    """
    Crea una imagen RGB de gradiente horizontal de 0 a 255.
    """
    gradiente = np.tile(np.linspace(0, 255, ancho, dtype=np.uint8), (alto, 1))
    imagen_rgb = np.stack([gradiente]*3, axis=2)  # Convertir a RGB replicando el canal
    return imagen_rgb

# Crear la imagen sintética
imagen_sintetica = crear_imagen_gradiente()

# Mostrar información
mostrar_info_imagen(imagen_sintetica, "Imagen Sintética")

# Mostrar la imagen (opcional)
plt.imshow(imagen_sintetica)
plt.title("Imagen Sintética (Gradiente)")
plt.axis("off")
plt.show()
````

````python
# Aplicá segmentación por umbralización
# Probá diferentes valores hasta encontrar el mejor

# Tu código aquí
import cv2
import matplotlib.pyplot as plt
import numpy as np

# Leer la imagen en escala de grises
ruta_img = "/content/drive/MyDrive/dataset/buenacalidad.jpg"
img_bgr = cv2.imread(ruta_img)
img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)

# 1. Umbralización binaria global con diferentes umbrales
_, thresh_127 = cv2.threshold(img_gray, 127, 255, cv2.THRESH_BINARY)
_, thresh_150 = cv2.threshold(img_gray, 150, 255, cv2.THRESH_BINARY)
_, thresh_180 = cv2.threshold(img_gray, 180, 255, cv2.THRESH_BINARY)

# 2. Umbralización adaptativa (media y gaussiana)
adaptive_mean = cv2.adaptiveThreshold(img_gray, 255, cv2.ADAPTIVE_THRESH_MEAN_C,
                                      cv2.THRESH_BINARY, blockSize=11, C=2)

adaptive_gaussian = cv2.adaptiveThreshold(img_gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
                                          cv2.THRESH_BINARY, blockSize=11, C=2)

# Mostrar todas las imágenes
titulos = [
    "Original (Gris)",
    "Global Thresh 127",
    "Global Thresh 150",
    "Global Thresh 180",
    "Adaptativa - Media",
    "Adaptativa - Gaussiana"
]

imagenes = [
    img_gray,
    thresh_127,
    thresh_150,
    thresh_180,
    adaptive_mean,
    adaptive_gaussian
]

plt.figure(figsize=(15, 8))
for i in range(len(imagenes)):
    plt.subplot(2, 3, i+1)
    cmap = 'gray' if i > 0 else 'gray'
    plt.imshow(imagenes[i], cmap=cmap)
    plt.title(titulos[i])
    plt.axis('off')
plt.tight_layout()
plt.show()
````

````python
# Y la aplicamos con warpPerspective
# El segundo parametro es esta matriz y el tercero debe ser una tupla con especificando (ancho, alto)
imagen3_output = cv2.warpPerspective(imagen3, matrix, (w2, h2))
plt.imshow(imagen3_output);
````

````python
# Convertir a escala de grises
imagen3_gray = cv2.cvtColor(imagen3_output, cv2.COLOR_RGB2GRAY)

# Ecualizar histograma para mejorar contraste
imagen3_eq = cv2.equalizeHist(imagen3_gray)

# Mostrar imagen mejorada
plt.imshow(imagen3_eq, cmap='gray')
plt.title("Transformada + Ecualizada")
plt.axis('off')
plt.show()
````
