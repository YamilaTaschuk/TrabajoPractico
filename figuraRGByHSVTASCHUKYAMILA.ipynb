```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

fig, ax = plt.subplots(figsize=(4, 4))
ax.set_xlim(0, 400)
ax.set_ylim(0, 400)
ax.set_aspect('equal')
ax.axis('off')  # Ocultar ejes

# Rectángulo rojo con RGB
rect = patches.Rectangle((100, 100), 200, 200, facecolor=(1, 0, 0))  # Rojo en RGB
ax.add_patch(rect)

plt.savefig("figura_rgb.jpg", bbox_inches='tight')
plt.show()

```

Ejercicio 2: Figura en Espacio de Color HSV

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import colorsys

# Convertir HSV a RGB (Hue entre 0-1, Saturación y Valor entre 0-1)
hsv_color = (200 / 360, 0.8, 1.0)  # Hue=200°, Saturación=80%, Brillo=100%
rgb_color = colorsys.hsv_to_rgb(*hsv_color)

fig, ax = plt.subplots(figsize=(4, 4))
ax.set_xlim(0, 400)
ax.set_ylim(0, 400)
ax.set_aspect('equal')
ax.axis('off')

# Círculo con color HSV convertido a RGB
circle = patches.Circle((200, 200), 100, facecolor=rgb_color)
ax.add_patch(circle)

plt.savefig("figura_hsv.jpg", bbox_inches='tight')
plt.show()
```
