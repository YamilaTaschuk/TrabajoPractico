## Paso 2: Código base del clasificador

Este es el código completo. Tu tarea es modificar solo la línea que dice `categorias_candidatas`.

````python
from transformers import pipeline
from PIL import Image
import requests
from io import BytesIO
import gradio as gr

# Cargar modelo CLIP
print("Cargando modelo CLIP...")
clasificador = pipeline(
    "zero-shot-image-classification",
    model="openai/clip-vit-base-patch32"
)
print("Modelo cargado correctamente")

# MODIFICÁ ESTA LÍNEA SEGÚN LA TAREA QUE ELIJAS
categorias_candidatas = [
    "playa",
    "montaña",
    "bosque",
    "desierto"
]

def clasificar_imagen(imagen):
    """
    Clasifica una imagen según las categorías definidas.
    """
    resultados = clasificador(imagen, candidate_labels=categorias_candidatas)

    # Formatear resultados para Gradio
    return {
        resultado['label']: float(resultado['score'])
        for resultado in resultados
    }

# Crear interfaz web
demo = gr.Interface(
    fn=clasificar_imagen,
    inputs=gr.Image(type="pil", label="Subí una imagen"),
    outputs=gr.Label(num_top_classes=4, label="Resultados"),
    title="Clasificador de Paisajes con CLIP",
    description="Este clasificador usa CLIP para identificar tipos de paisajes en imágenes",
    examples=[
        ["https://images.unsplash.com/photo-1507525428034-b723cf961d3e?ixlib=rb-4.1.0&auto=format&fit=crop&q=80&w=1169"],  # Imagen de playa
    ],
)

# Lanzar aplicación
demo.launch()
````

````python
from transformers import pipeline
from PIL import Image
import requests
from io import BytesIO
import gradio as gr

# Cargar modelo CLIP
print("Cargando modelo CLIP...")
clasificador = pipeline(
    "zero-shot-image-classification",
    model="openai/clip-vit-base-patch32"
)
print("Modelo cargado correctamente")

# Categorías para calidad de frutas
categorias_candidatas = [
    "fruta fresca y madura",
    "fruta podrida o en mal estado",
    "fruta verde sin madurar",
    "fruta procesada o cortada"
]

def clasificar_imagen(imagen):
    resultados = clasificador(imagen, candidate_labels=categorias_candidatas)
    return {resultado['label']: float(resultado['score']) for resultado in resultados}

demo = gr.Interface(
    fn=clasificar_imagen,
    inputs=gr.Image(type="pil", label="Subí una imagen"),
    outputs=gr.Label(num_top_classes=4, label="Resultados"),
    title="Detector de Calidad de Alimentos con CLIP",
    description="Clasifica frutas según su estado usando el modelo CLIP",
)

demo.launch()
````

¿Cómo cambió el comportamiento del modelo?

CLIP se adapta muy bien a nuevas tareas gracias a las descripciones en lenguaje natural.

Las predicciones cambian según el contexto que le das en categorias_candidatas.

¿Qué tan precisas son las predicciones?

Puede ser sorprendentemente preciso con imágenes claras y bien iluminadas.

Puede fallar si las imágenes son confusas, borrosas o muestran múltiples objetos o personas con contextos mixtos.

Limitaciones del modelo:

No entiende contexto complejo (por ejemplo, diferencias sutiles entre "feliz" y "neutral").

Puede confundirse si las etiquetas son muy similares entre sí.

No es lo mismo que un modelo entrenado específicamente con ejemplos etiquetados.

¿Qué imágenes funcionan mejor?

Imágenes limpias, con el objeto/persona en primer plano.

Buena iluminación y resolución.

Sin mucho ruido de fondo.

    """

### Paso 3.2 Alternativo: Transfer Learning Avanzado (Opcional)

Si querés más control sobre el proceso de entrenamiento, podés usar Transfer Learning con código en lugar de Teachable Machine. Esta opción requiere más conocimiento técnico pero te da mayor flexibilidad.

**Ventajas del Transfer Learning avanzado**:

**Código de ejemplo para Transfer Learning con MobileNetV2**:

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense
from tensorflow.keras.models import Model

# 1. Configurar Data Augmentation
datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=10,
    width_shift_range=0.15,
    height_shift_range=0.15,
    shear_range=5,
    zoom_range=[0.7, 1.3],
    validation_split=0.2
)

# 2. Cargar datos
data_gen_entrenamiento = datagen.flow_from_directory(
    "dataset/",
    target_size=(224, 224),
    batch_size=32,
    shuffle=True,
    subset="training"
)

data_gen_validacion = datagen.flow_from_directory(
    "dataset/",
    target_size=(224, 224),
    batch_size=32,
    shuffle=True,
    subset="validation"
)

# 3. Cargar modelo base preentrenado
mobilenetv2 = MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights='imagenet'
)

# Congelar capas del modelo base
mobilenetv2.trainable = False

# 4. Construir modelo completo
input_layer = tf.keras.Input(shape=(224, 224, 3))
x = mobilenetv2(input_layer)
x = GlobalAveragePooling2D()(x)
output_layer = Dense(
    data_gen_entrenamiento.num_classes,
    activation="softmax"
)(x)

modelo = Model(inputs=input_layer, outputs=output_layer)

# 5. Compilar
modelo.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

# 6. Entrenar
EPOCAS = 20
historial = modelo.fit(
    data_gen_entrenamiento,
    epochs=EPOCAS,
    validation_data=data_gen_validacion
)

# 7. Guardar modelo
modelo.save("mi_modelo_transfer_learning.h5")

# Guardar también las etiquetas
with open("labels.txt", "w") as f:
    for clase, indice in data_gen_entrenamiento.class_indices.items():
        f.write(f"{indice} {clase}\n")

print("Modelo entrenado y guardado correctamente")
```
