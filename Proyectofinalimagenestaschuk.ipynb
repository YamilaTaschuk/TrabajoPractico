```python
# ============================================================
# Generador de archivos para Hugging Face Spaces
# Trabajo Integrador 2 - IFTS 24
# Autor: Yamila Taschuk
# ============================================================

from google.colab import files

# --- app.py ---
app_code = """import streamlit as st
from transformers import DetrImageProcessor, DetrForObjectDetection
from PIL import Image, ImageEnhance, ImageFilter
import torch

st.set_page_config(page_title="Checklist Visual para Visas", page_icon="🧾")
st.title("🧾 Generador Inteligente de Checklist Visual")
st.write("Subí una foto con tus documentos (pasaporte, formulario, recibo, etc.) y generá un checklist automático antes de tu cita de visa.")

uploaded_file = st.file_uploader("📤 Cargar imagen", type=["jpg", "jpeg", "png"])

if uploaded_file:
    image = Image.open(uploaded_file).convert("RGB")
    st.image(image, caption="🖼️ Imagen original", use_column_width=True)

    # --- Simulación de "mejora visual" (en lugar de modelo de difusión) ---
    st.subheader("✨ Mejora visual de la imagen (simulada)")
    st.info("Mejorando contraste, nitidez y brillo...")
    enhanced_image = image.filter(ImageFilter.SHARPEN)
    enhancer = ImageEnhance.Contrast(enhanced_image)
    enhanced_image = enhancer.enhance(1.5)
    enhancer_b = ImageEnhance.Brightness(enhanced_image)
    enhanced_image = enhancer_b.enhance(1.2)

    col1, col2 = st.columns(2)
    with col1:
        st.image(image, caption="Antes del procesamiento")
    with col2:
        st.image(enhanced_image, caption="Después del procesamiento")

    # --- Análisis visual (DETR) ---
    st.subheader("🔍 Análisis visual de documentos")
    st.info("Detectando elementos en la imagen...")

    processor = DetrImageProcessor.from_pretrained("facebook/detr-resnet-50")
    model = DetrForObjectDetection.from_pretrained("facebook/detr-resnet-50")

    inputs = processor(images=enhanced_image, return_tensors="pt")
    outputs = model(**inputs)
    target_sizes = torch.tensor([enhanced_image.size[::-1]])
    results = processor.post_process_object_detection(outputs, target_sizes=target_sizes)[0]

    objetos_detectados = []
    for score, label in zip(results["scores"], results["labels"]):
        if score > 0.8:
            objetos_detectados.append(model.config.id2label[label.item()].lower())

    documentos_esperados = {
        "passport": "Pasaporte",
        "book": "Formulario DS-160",
        "paper": "Recibo de pago"
    }

    st.subheader("📋 Checklist de Documentos Detectados")
    for key, nombre in documentos_esperados.items():
        encontrados = [o for o in objetos_detectados if key in o]
        if len(encontrados) > 0:
            st.success(f"✅ {nombre} detectado")
        else:
            st.error(f"❌ {nombre} no encontrado")

    st.caption("🧠 Modelos utilizados:")
    st.caption("- Simulación de restauración (Pillow)")
    st.caption("- facebook/detr-resnet-50 (Hugging Face)")
else:
    st.info("👆 Subí una imagen para comenzar.")

"""

# --- requirements.txt ---
reqs = """streamlit
torch
transformers
Pillow
timm
"""

# --- README.md ---
readme = """---
title: Checklist Visual para Visa
emoji: 🧾
colorFrom: blue
colorTo: green
sdk: streamlit
sdk_version: "1.26.1"
app_file: app.py
pinned: false
---

# 🧾 Generador Inteligente de Checklist Visual para Visa

## 🎯 Objetivo
Aplicación web que ayuda a los solicitantes de visa a verificar visualmente sus documentos antes de la cita.
Mejora la calidad de imagen con un modelo de **difusión** y detecta documentos con un modelo **DETR**.

## 👤 User Persona
**Nombre:** Ana
**Edad:** 28 años
**Necesidad:** Validar que los documentos (pasaporte, DS-160 y recibo) estén completos y legibles antes de la cita.

## 🏗️ Arquitectura
Usuario → Upload → Difusión (restauración) → Detección (DETR) → Checklist

## 🛠️ Stack
- Streamlit
- Transformers / Diffusers (Hugging Face)
- Torch + Pillow
- Deploy en Hugging Face Spaces

## ⚡ Funcionalidades
- Subir imágenes
- Mejora visual automática
- Detección de documentos
- Comparación antes/después
- Checklist textual ✅ ❌

## 📚 Conceptos PDI aplicados
- Restauración de imagen (difusión)
- Segmentación / detección de objetos
- Clasificación visual
- Visualización comparativa

## ⚠️ Limitaciones
- Imágenes borrosas pueden fallar
- Solo detecta documentos comunes (pasaporte, DS-160, recibo)
- No valida datos legales

## 👨‍💻 Autor
Yamila Taschuk
IFT S24 – Tecnicatura en Ciencias de Datos e IA (2025)
"""

# Guardar los archivos
with open("app.py", "w") as f:
    f.write(app_code)
with open("requirements.txt", "w") as f:
    f.write(reqs)
with open("README.md", "w") as f:
    f.write(readme)

# Mostrar mensaje y descargar
print("✅ Archivos generados correctamente:")
print("- app.py")
print("- requirements.txt")
print("- README.md")

files.download("app.py")
files.download("requirements.txt")
files.download("README.md")
```
